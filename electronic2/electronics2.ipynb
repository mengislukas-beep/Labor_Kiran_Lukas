```python
def f(omega, a, b):
    return 1 / np.sqrt(1 + 1000**2 * (omega*b - 1/(omega*a))**2)
```

```python
##### 3.2.3
# to do

L = 0.01
C = 100 * 10**(-9)
R = 1000
U_in = 1



# Q = omega_0 / d_omega
# d_omega = omega_o - omega_u
# are values where current / voltage has dropped by 1 / sqrt(2)



wdata = np.array([
    10000, 15000, 20000, 25000, 30000,
    31623, 35000, 40000, 45000, 50000,
    60000, 70000
])

Adata = np.array([
    0.110, 0.187, 0.316, 0.539, 0.948,
    1.000, 0.736, 0.496, 0.369, 0.287,
    0.205, 0.159
])

popt, pcov = curve_fit(
    f, wdata, Adata,
    p0=[L, C]
)

a,b = popt 
print('theoretical Q =', R * np.sqrt(C / L))

print(a,b)

plt.plot(wdata, Adata)
plt.plot(wdata, f(wdata, a, b))
```
