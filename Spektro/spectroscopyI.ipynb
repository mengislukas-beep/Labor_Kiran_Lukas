```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
p_cent = 720
k0 = 950 #+- 1
k_cent = np.array([1070, 1072, 1101, 1131, 1310, 849, 800])
lamda_cent = np.array([505, 492, 470, 447, 389, 668, 706])
```

```python
def lamda_cent_fit (k, bb, aa):
    return bb * k + aa
```

```python
from scipy.optimize import curve_fit
popt, pcov = curve_fit(lamda_cent_fit, k_cent, lamda_cent, absolute_sigma=True)
b = popt[0]
a = popt[1]
print(b)
print(a)

plt.plot(k_cent, lamda_cent, linestyle='', marker='o')
plt.plot(k_cent, lamda_cent_fit(k_cent, b, a))
```

```python
k_fix = 1066
p_k = np.array([470, 800, 925])
lamda_k = np.array([492, 501, 505])
```

```python
def lamda_fit (p, dd):
    return b * k_fix + a + (p - p_cent) / dd
```

```python
from scipy.optimize import curve_fit
popt, pcov = curve_fit(lamda_fit, p_k, lamda_k, absolute_sigma=True)
d = popt[0]
print(d)
```

```python
plt.plot(p_k, lamda_k, linestyle='', marker='o')
plt.plot(p_k, lamda_fit(p_k, d))
plt.plot(p_cent, lamda_fit(p_cent, d), marker='o', color='r')
```

```python
def lamda (p, k):
    return b * k + a + (p - p_cent) / d
```
