```python
##### 2.1
U0 = 10 #V
R1 = 4700
R2 = 10000
print("expected U0 =", (R2 * U0) / (R1 + R2))

R1 = R1 * 1000
R2 = R2 * 1000
print("expected U0 =", (R2 * U0) / (R1 + R2))

Uout = 5.33 #measurement
R_ersatz = R1 / U0 * 1 / (1 / Uout - 1 / U0)
R_i = 1 / (1 / R_ersatz - 1 / R2)
print("calculated R_intern =", R_i)
print(R1 * R2 * Uout / (R2 * (U0 - Uout) -R1 * Uout) / 10**6)
```
